# Caso practico

Actividades a realizar

1. Construcción de un modelo de regresión polinomial y su evaluación sobre el conjunto test. Para determinar el grado de la transformación polinomial utiliza un conjunto de validación sobre los siguientes valores de grado de polinomio: [2, 3]. 

2. Construcción de un modelo de regresión Ridge y su evaluación sobre el conjunto test. Para determinar el valor del hiperparámetro de regularización utiliza el método GridSearchCV sobre los siguientes valores para α: [0.1, 0.5, 1.0]. Utiliza para la búsqueda la métrica RMSE.

3. Construcción de un modelo de regresión Lasso y su evaluación sobre el conjunto test. Para determinar el valor del hiperparámetro de regularización utiliza el método GridSearchCV sobre los siguientes valores para α: [1, 2, 3, 4, 5]. Utiliza para la búsqueda la métrica RMSE.

4. Elaboración de una tabla comparativa mostrando el rendimiento sobre test de los tres modelos seleccionados (con mejores rendimientos) de las actividades 1, 2 y 3 y el de regresión lineal construido en el módulo anterior.

5. Utilizando el modelo Lasso determina cuáles fueron las variables que se consideraron importantes para la predicción.

Para la construcción del modelo de regresión polinomial puedes utilizar el notebook “La regresión polinomial en scikit-learn: caso con múltiples variables”. Para incluir la búsqueda de hiperparámetros puedes basarte en el notebook “Regresión regularizada. Caso Ridge” y utilizar la misma sección de código relacionada con la búsqueda del parámetro α con un conjunto de validación, pero ahora para la búsqueda del mejor grado de polinomio. 

Para la construcción de los modelos de regresión regularizada, Ridge y Lasso, puedes utilizar los notebooks “Regresión regularizada. Caso Ridge” y “Regresión regularizada. Caso Lasso”. Toma en cuenta que cuando invoques a la función GridSearchCV uno de los parámetros que debes indicarle es scoring = 'neg_root_mean_squared_error', con el fin de que utilice RMSE como métrica de evaluación en la búsqueda.

Considera las etapas de la metodología ASUM-DM para el desarrollo de proyectos de analítica y sigue el proceso recomendado, desde la carga de los datos hasta la evaluación e interpretación del modelo. Muy importante es que en la etapa de preparación de datos sigas los mismos pasos que aplicaste para el caso de regresión lineal.
